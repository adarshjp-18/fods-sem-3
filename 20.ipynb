# Exp No:20 Experiment to understand K-means clustering algorithm for a given data set
**Description:** Understand the K-means clustering algorithm for the dataset given.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
df = pd.read_csv(r"C:\Users\prasa\Downloads\Mall_Customers - Mall_Customers.csv")
df.info()
df.head()
```

```python
sns.pairplot(df)
```

```python
features=df.iloc[:,[3,4]].values
```

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=5,n_init=10)
model.fit(features)
```

```python
Final=df.iloc[:,[3,4]].copy()
Final['label']=model.predict(features)
Final.head()
```

```python
sns.set_style("whitegrid")
sns.FacetGrid(Final,hue="label",height=8) \
.map(plt.scatter,"Annual Income (k$)", "Spending Score (1-100)") \
.add_legend();
plt.show()
```

```python
features_el=df.iloc[:,[2,3,4]].values
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,10):
    model=KMeans(n_clusters=i,n_init=10)
    model.fit(features_el)
    wcss.append(model.inertia_)
plt.plot(range(1,10),wcss)
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```
